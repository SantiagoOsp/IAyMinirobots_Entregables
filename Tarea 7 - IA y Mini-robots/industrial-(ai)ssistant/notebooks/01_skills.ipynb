## Definiciones propuestas y criterios de revisión

Las definiciones siguientes complementan la respuesta. Deben compararse con los `SKILL.md` que ya tienes; no certifican su contenido ni los reemplazan.

### Documentación de tickets

**Activación:** cuando se solicite documentar un trabajo o falla. **Método:** identificar contexto y hechos; separar inferencias; registrar datos pendientes; definir alcance y criterios comprobables. **Salida:** título, problema, resultado esperado, alcance, fuera de alcance, criterios, dependencias, riesgos y evidencias. **Regla:** no convertir una hipótesis de causa en diagnóstico confirmado.

### Implementación, código y pruebas

**Activación:** ante un ticket de software y un proyecto. **Método:** leer el ticket y código, diseñar el cambio acotado, ejecutar pruebas pertinentes y registrar resultados reales. **Salida:** cambios, archivos, pruebas, criterios verificados y limitaciones. **Regla:** no declarar éxito con pruebas fallidas ni eliminar pruebas para ocultar errores.

### Explicación HTML

**Activación:** cuando se solicite documentar una implementación realizada. **Método:** revisar cambios y evidencias; describir problema, arquitectura, código relevante y verificación. **Salida:** `report.html` con HTML5 y CSS incorporado. **Regla:** distinguir resultados observados de ejemplos y no depender de servicios externos para presentar la explicación.

### Lectura de requerimientos

**Activación:** ante documentos y una solicitud de extraer requisitos. **Método:** localizar requerimientos explícitos, restricciones, interfaces y ambigüedades. **Salida:** tabla `ID | tipo | requerimiento | fuente | estado`. **Regla:** cada requisito debe señalar su fuente; las inferencias y contradicciones deben identificarse.

## Verificar tus cuatro archivos

```python
from pathlib import Path
import sys

# Coloca los notebooks en la raíz del proyecto o en su carpeta notebooks/.
# Si están en otro lugar, escribe aquí la ruta de tu proyecto.
PROJECT_DIR = None  # Ejemplo: Path(r"C:\industrial-ai-assistant")

if PROJECT_DIR is None:
    candidates = [Path.cwd(), *Path.cwd().parents]
    PROJECT_DIR = next(
        (p for p in candidates if (p / "config.py").is_file()
         and (p / "app.py").is_file()), None
    )
if PROJECT_DIR is None:
    raise FileNotFoundError("Indica PROJECT_DIR con la ruta de tu proyecto.")
PROJECT_DIR = Path(PROJECT_DIR).expanduser().resolve()
if not (PROJECT_DIR / "config.py").is_file():
    raise FileNotFoundError("La ruta indicada no contiene config.py.")
if str(PROJECT_DIR) not in sys.path:
    sys.path.insert(0, str(PROJECT_DIR))
print("Proyecto:", PROJECT_DIR)
```

```python
skill_names = ["ticket-documentation", "implementation", "explanation", "requirements-reader"]
skill_texts = {}
for name in skill_names:
    path = PROJECT_DIR / "skills" / name / "SKILL.md"
    if path.is_file():
        skill_texts[name] = path.read_text(encoding="utf-8")
        print(name, "— archivo disponible:", len(skill_texts[name]), "caracteres")
    else:
        print(name, "— PENDIENTE: revisa nombre y ubicación:", path)
```

```python
# Vista previa de las instrucciones reales, sin ejecutar código generado.
for name, text in skill_texts.items():
    print("\nSKILL:", name)
    print(text[:1500])
```
